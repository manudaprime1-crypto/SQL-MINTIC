# ANÁLISIS FINANCIERO

```python
import yfinance
```

```python
codigo = input("Escriba el código de la acción: ")

data = yfinance.download(["codigo"], period="6mo")

cierre = data.Close["codigo"]
```

```python
cierre.plot()
```

```python
maxima = (round(cierre.max(), 1))
minima = (round(cierre.min(), 1))
valor_medio = (round(cierre.mean(), 1))
```

```python
print(maxima)
print(minima)
print(valor_medio)
```
